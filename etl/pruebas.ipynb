```python
from extract import extract
from transform import transform
```

```python
df = extract("../data/raw/temperature_observations_colombia.csv")

print("Shape del dataset original:")
print(df.shape)
```

```python
dim_fecha, dim_estacion, dim_sensor, fact = transform(df)
```

```python
# tamanios de las tablas
print("dim_fecha:", dim_fecha.shape)
print("dim_estacion:", dim_estacion.shape)
print("dim_sensor:", dim_sensor.shape)
print("fact:", fact.shape)
```

```python
# validacion de nulos
print("Nulos en dimensiones:")

print(dim_fecha.isnull().sum())
print(dim_estacion.isnull().sum())
print(dim_sensor.isnull().sum())

print("\nNulos en fact:")
print(fact.isnull().sum())
```

```python
# Validamos que las FK existen
print("estaciones en fact que no existen en dim_estacion:")
print(fact["estacion_key"].isin(dim_estacion["estacion_key"]).value_counts())

print("sensores en fact que no existen en dim_sensor:")
print(fact["sensor_key"].isin(dim_sensor["sensor_key"]).value_counts())
```
